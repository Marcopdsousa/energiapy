## Problem Statement

Small energy systems example to learn energiapy.

## Nomenclature

*Sets*

$\mathcal{L}$ - set of all locations l

$\mathcal{P}$ - set of all processes p

$\mathcal{R}$ - set of all resources r

$\mathcal{M}$ - set of all materials m

$\mathcal{T}^{net}$ - set of temporal periods t for network-level decision making

$\mathcal{T}^{sch}$ - set of temporal periods t for schedule-level decision making


*Subsets*

$\mathcal{R}^{storage}$ - set of resources that can be stored

$\mathcal{R}^{cons}$ - set of resources that can be consumed

$\mathcal{R}^{demand}$ - set of resources that satisfy demand


*Continuous Variables*

$P_{l,p,t}$ - production level of process $p \in \mathcal{P}$ in time period $t \in \mathcal{T}^{sch}$

$C_{l,r,t}$ - consumption of resource $r \in \mathcal{R}^{cons}$ in time period $t \in \mathcal{T}^{sch}$

$S_{l,r,t}$ - supplied or sold amount of resource $r \in \mathcal{R}^{demand}$ in time period $t \in \mathcal{T}^{sch}$

$Inv_{l,r,t}$ - inventory level of resource $r \in \mathcal{R}^{storage}$ in time period $t \in \mathcal{T}^{sch}$

$Cap^S_{l,r,t}$ - installed inventory capacity for resource $r \in \mathcal{R}^{storage}$ in time period $t \in \mathcal{T}^{net}$

$Cap^P_{l,p,t}$ - installed production capacity for process $p \in \mathcal{P}$ in time period $t \in \mathcal{T}^{net}$

$Mat_{l,p,m,t}$ - material $m \in \mathcal{M}$ used by process $p \in \mathcal{P}$ in time period $t \in \mathcal{T}^{net}$

$Em^r_{l,r,t}$ - emissions associated with resource consumption in time period $t \in \mathcal{T}^{sch}$

$Em^m_{l,m,t}$ - emissions associated with material use in time period $t \in \mathcal{T}^{net}$

$Em^p_{l,p,t}$ - emissions associated with process capacity in time period $t \in \mathcal{T}^{net}$


*Parameters*

$Cap^{P-max}_{l,p,t}$ - maximum production capacity of process $p \in \mathcal{P}$ in time period $t \in \mathcal{T}^{net}$

$Cap^{S-max}_{l,r,t}$ - maximum inventory capacity for resource $r \in \mathcal{R}^{storage}$ in time period $t \in \mathcal{T}^{net}$

$Capex_{l,p,t}$ - capital expenditure for process $p \in \mathcal{P}$ in time period $t \in \mathcal{T}^{net}$

$Fopex_{l,p,t}$ - fixed operating expenditure for process $p \in \mathcal{P}$ in time period $t \in \mathcal{T}^{net}$

$Vopex_{l,r,t}$ - variable operating expenditure coefficient for resource or production term in time period $t$

$Price_{l,r,t}$ - purchase price for resource $r \in \mathcal{R}^{cons}$ in time period $t \in \mathcal{T}^{sch}$

$C^{max}_{l,r,t}$ - maximum consumption availability for resource $r \in \mathcal{R}^{cons}$ in time period $t \in \mathcal{T}^{sch}$

$D_{l,r,t}$ - demand for resource $r \in \mathcal{R}^{demand}$ in time period $t \in \mathcal{T}^{sch}$

$\eta(p,r)$ - conversion factor from process $p \in \mathcal{P}$ to resource $r \in \mathcal{R}$

$Mat^{cons}_{p,m}$ - material $m \in \mathcal{M}$ consumed by process $p \in \mathcal{P}$

$GWP^r_{l,r,t}$ - global warming indicator for resource consumption in time period $t$

$GWP^m_{l,m,t}$ - global warming indicator for material use in time period $t$

$GWP^p_{l,p,t}$ - global warming indicator for process capacity in time period $t$

$\alpha$ - annualization factor

## LP Formulation

Given is a multi-scale modeling and optimization LP framework for the simultaneous design and scheduling of a single-location energy system.

\begin{align}
\min \quad
& \sum_{l \in \mathcal{L}} \Bigg(
    \sum_{t \in \mathcal{T}^{net}} \sum_{p \in \mathcal{P}}
    \left( \alpha \, Capex_{l,p,t} + Fopex_{l,p,t} \right) Cap^P_{l,p,t}
    \notag \\
& \qquad\qquad +
    \sum_{t \in \mathcal{T}^{sch}} \sum_{r \in \mathcal{R}}
    Vopex_{l,r,t} \, P_{l,r,t}
    +
    \sum_{t \in \mathcal{T}^{sch}} \sum_{r \in \mathcal{R}^{cons}}
    C_{l,r,t} \, Price_{l,r,t}
\Bigg)
\end{align}

\begin{align}
Cap^S_{l,r,t}
&\leq Cap^{S\text{-}max}_{l,r,t}
&& \forall r \in \mathcal{R}^{storage},\; t \in \mathcal{T}^{net} \\
Cap^P_{l,p,t}
&\leq Cap^{P\text{-}max}_{l,p,t}
&& \forall p \in \mathcal{P},\; t \in \mathcal{T}^{net},\; l \in \mathcal{L} \\
P_{l,p,t}
&\leq Cap^P_{l,p,t}
&& \forall p \in \mathcal{P},\; t \in \mathcal{T}^{sch} \\
Inv_{l,r,t}
&\leq Cap^S_{l,r,t}
&& \forall r \in \mathcal{R}^{storage},\; t \in \mathcal{T}^{sch} \\
S_{l,r,t}
&\geq D_{l,r,t}
&& \forall r \in \mathcal{R},\; t \in \mathcal{T}^{sch} \\
C_{l,r,t}
&\leq C^{max}_{l,r,t}
&& \forall r \in \mathcal{R},\; t \in \mathcal{T}^{sch} \\
\sum_{p \in \mathcal{P}} P_{l,p,t}\,\eta(p,r) + C_{l,r,t} + Inv_{l,r,t-1}
&= Inv_{l,r,t} + S_{l,r,t}
&& \forall r \in \mathcal{R}^{cons},\; t \in \mathcal{T}^{sch},\; l \in \mathcal{L}
\end{align}

\begin{align}
Mat_{l,p,m,t}
&= Mat^{cons}_{p,m}\, Cap^P_{l,p,t}
&& \forall m \in \mathcal{M},\; p \in \mathcal{P},\; t \in \mathcal{T}^{net} \\
Em^{r}_{l,r,t}
&= GWP^{r}_{l,r,t}\, C_{l,r,t}
&& \forall r \in \mathcal{R},\; t \in \mathcal{T}^{sch} \\
Em^{m}_{l,m,t}
&= GWP^{m}_{l,m,t} \sum_{p \in \mathcal{P}} Mat_{l,p,m,t}
&& \forall m \in \mathcal{M},\; t \in \mathcal{T}^{net},\; l \in \mathcal{L} \\
Em^{p}_{l,p,t}
&= GWP^{p}_{l,p,t}\, Cap^P_{l,p,t}
&& \forall p \in \mathcal{P},\; t \in \mathcal{T}^{net},\; l \in \mathcal{L}
\end{align}

\begin{equation}
S_{l,r,t},\; C_{l,r,t},\; Inv_{l,r,t},\; P_{l,p,t},\; Cap^P_{l,p,t},\; Cap^S_{l,r,t},\; Mat_{l,p,m,t},\; Em^{p/r/m}_{l,p/r/m,t} \in \mathbb{R}_{\ge 0}
\end{equation}

```python
# demand_factor = pandas.DataFrame(data={'Power': [0.6, 0.7, 0.8, 0.3]})
# capacity_factor_pv = pandas.DataFrame(data={'PV': [0.6, 0.8, 0.9, 0.7]})
# capacity_factor_wf = pandas.DataFrame(data={'WF': [0.9, 0.8, 0.5, 0.7]})

import pandas as pd

demand_factor = pd.DataFrame(
    data={
        "Power": [0.6, 0.7, 0.8, 0.3]
    },
    index=pd.MultiIndex.from_product(
        [range(4)],
        names=["time"]
    )
)

demand_factor = demand_factor.to_dict()

# print(demand_factor)

capacity_factor_pv = pd.DataFrame(
    data={
        "PV": [0.6, 0.8, 0.9, 0.7]
    },
    index=pd.MultiIndex.from_product(
        [range(4)],
        names=["time"]
    )
)

capacity_factor_pv = capacity_factor_pv.to_dict()

# print(capacity_factor_pv_dict)


capacity_factor_wf = pd.DataFrame(
    data={
        "WF": [0.9, 0.8, 0.5, 0.7]
    },
    index=pd.MultiIndex.from_product(
        [range(4)],
        names=["time"]
    )
)

capacity_factor_wf = capacity_factor_wf.to_dict()

# print(capacity_factor_wf_dict)
```

## Declare Temporal Scale


```python
scales = TemporalScale(discretization_list=[1,4])
```

## Declare Resources

```python
Solar = Resource(name='Solar', cons_max=100, basis='MW', label='Solar Power')
Wind = Resource(name='Wind', cons_max=100, basis='MW', label='Wind Power')
Power = Resource(name='Power', basis='MW', demand=True, label='Power generated', varying=[VaryingResource.DETERMINISTIC_DEMAND])

#, varying=[VaryingResource.DETERMINISTIC_PRICE]
```

## Declare Processes

```python
LiI = Process(name='LiI', storage=Power, capex=1302182, fopex=41432, vopex=2000, prod_max = 100, prod_min=10, store_max=100, label='Lithium-ion battery', basis='MW')
WF = Process(name='WF', conversion={Wind: -1, Power: 1}, capex=990637, fopex=3354, vopex=4953, prod_max=100, prod_min=10, label='Wind mill array', varying=[VaryingProcess.DETERMINISTIC_CAPACITY])
PV = Process(name='PV', conversion={Solar: -1, Power: 1}, capex=567000, fopex=872046, vopex=90000, prod_max=100, prod_min=10, label='Photovoltaic array', varying=[VaryingProcess.DETERMINISTIC_CAPACITY])

#
#, varying=[VaryingProcess.DETERMINISTIC_CAPACITY]
#
```

## Declare Location

```python
place = Location(name='place', processes={LiI,PV,WF}, demand_factor={Power: demand_factor}, capacity_factor={PV: capacity_factor_pv, WF: capacity_factor_wf}, capacity_scale_level=1, demand_scale_level=1, scales=scales, label='some place')

#, capacity_factor={PV: capacity_factor_pv, WF: capacity_factor_wf}
```

## Declare Scenario

```python
case = Scenario(name='case', network=place, demand_scale_level=1, purchase_scale_level=1, network_scale_level=0, scheduling_scale_level=1, scales=scales, demand={place: {Power: 180}}, label='small energy system')
```
